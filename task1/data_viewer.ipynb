```python
from torchvision.datasets import STL10
import matplotlib.pyplot as plt
import random

dataset = STL10(root="../data", split="test", download=False)
rng = random.Random(6304)
images_per_class = 36

for class_name in dataset.classes:
    indices = [
        i for i, label in enumerate(dataset.labels)
        if dataset.classes[label] == class_name
    ]

    selected = rng.sample(indices, min(images_per_class, len(indices)))

    fig, axes = plt.subplots(6, 6, figsize=(12, 12))
    fig.suptitle(f"{class_name.upper()} — dataset indices", fontsize=16)

    for ax, index in zip(axes.flat, selected):
        image, _ = dataset[index]
        ax.imshow(image)
        ax.set_title(f"idx={index}", fontsize=7)
        ax.axis("off")

    for ax in axes.flat[len(selected):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```
